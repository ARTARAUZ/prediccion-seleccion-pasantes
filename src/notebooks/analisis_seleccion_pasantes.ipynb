##  Antecedentes

En el ámbito empresarial y académico, el proceso de selección de pasantes constituye un reto operativo significativo. Las organizaciones suelen recibir un alto volumen de postulantes, lo que dificulta la gestión eficiente de las solicitudes y la identificación de los candidatos más adecuados. Este proceso implica analizar múltiples variables, como el rendimiento académico, las habilidades técnicas, las competencias blandas y la experiencia extracurricular, lo que aumenta la complejidad de la toma de decisiones.

Uno de los principales problemas en la selección de candidatos radica en la dependencia de métodos tradicionales, que suelen basarse en revisiones manuales de currículums y evaluaciones subjetivas por parte de los reclutadores. Esto puede generar inconsistencias en la evaluación, introducir sesgos y reducir la objetividad del proceso. Asimismo, la falta de estandarización en los criterios de selección dificulta la comparación equitativa entre postulantes.

Adicionalmente, el tiempo requerido para evaluar a cada candidato puede ser considerable, especialmente cuando el número de postulaciones es elevado. Esto provoca retrasos en la toma de decisiones, afecta la experiencia de los aspirantes y puede ocasionar la pérdida de talento frente a otras organizaciones con procesos más ágiles y eficientes. En consecuencia, las empresas enfrentan la necesidad de optimizar sus procesos de selección para mejorar la precisión, reducir los tiempos y garantizar decisiones más justas y fundamentadas.

## Descripción del conjunto de datos

## Descripcion General

El dataset **Internship Selection Dataset** representa los perfiles de los estudiantes y los resultados de la selección para prácticas. Incluye el rendimiento académico, las habilidades técnicas, la experiencia en proyectos, los rasgos de comportamiento y los parámetros de evaluación relacionados con la contratación. El conjunto de datos está diseñado con relaciones realistas entre las características y una lógica de selección no lineal, lo que lo hace adecuado para el análisis de datos, la visualización y aplicaciones de aprendizaje automático, como la clasificación y el análisis de la importancia de las características.

## Caracteristicas del Dataset

El dataset consta de 21 columnas:

- `student_id`: Corresponde al identificador único de cada estudiante.
- `CGPA`: Evalua el rendimiento académico en una escala de 10 puntos.
- `skills_score`: Nivel de habilidades técnicas (1-10)
- `projects_count`: Número de proyectos académicos o personales completados.
- `internships_done`: Número de previos interships completados.
- `communication_score`: Habilidades de comunicación (1-10)
- `aptitude_score`: Habilidades de logica y razonamiento (1-10)
- `coding_test_score`: Resultados en las pruebas de programación (1-10)
- `resume_score`: Calidad y solidez del CV (1-10)
- `extracurricular`: Participación en actividades extracurriculars (Yes-No)
- `college_tier`: Categoría de la institución (Tier 1, Tier 2, Tier 3)
- `hackatons_participated`: Número de hackatons atendidas
- `certifications_count`: Número de certificaciones
- `linkedin_activity_score`: Nivel de participación en plataformas profesionales (1-10)
- `github_score`: Actividad y contribución en Github (1-10)
- `soft_skills_score`: Habilidades interpersonales y de trabajo en equipo (1-10)
- `interview_score`: Resultado en las rondas de entrevista (1-10)
- `consistency_score`: Consistencia en resultados sobre el tiempo (1-10)
- `backlogs`: Número de asignaturas pendientes
- `placement_training`: Participación en programas de formación para la inserción laboral (Sí/No).
- `selected`: Indica si fue seleccionado o no. **Variable Objetivo**
  
## Tamaño del Dataset

- **Instancias:** 10000 registros.  
- **Atributos:** 21 características (20 predictoras y 1 variable objetivo).

#### Transformadores propios

```python
from sklearn.preprocessing import FunctionTransformer
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.metrics.pairwise import rbf_kernel
from sklearn.cluster import KMeans
```

```python
class ClusterSimilarity(BaseEstimator, TransformerMixin):
    def __init__(self, n_clusters=10, gamma=1.0, random_state=None):
        self.n_clusters = n_clusters
        self.gamma = gamma
        self.random_state = random_state
 
    def fit(self, X, y=None, sample_weight=None):
        self.kmeans_ = KMeans(self.n_clusters, n_init=10,
                              random_state=self.random_state)
        self.kmeans_.fit(X, sample_weight=sample_weight)
        return self  # always return self!
 
    def transform(self, X):
        return rbf_kernel(X, self.kmeans_.cluster_centers_, gamma=self.gamma)
    def get_feature_names_out(self, names=None):
        return [f"Cluster {i} similarity" for i in range(self.n_clusters)]
```
